```python
import random

import chess

from board_viewer import watch_agent_vs_agent
from nnues.agent import get_move as nnue_768_agent
from nnues.agent_1024 import get_move as nnue_1024_agent


def random_position(min_plies=8, max_plies=24):
    while True:
        board = chess.Board()
        for _ in range(random.randint(min_plies, max_plies)):
            if board.is_game_over():
                break
            board.push(random.choice(list(board.legal_moves)))
        if not board.is_game_over():
            return board.fen()


start_fen = random_position()
print('White: NNUE 1024 | Black: NNUE 768')
print('Starting FEN:', start_fen)
watch_agent_vs_agent(
    nnue_1024_agent,
    nnue_768_agent,
    fen_start=start_fen,
    pause_seconds=0,
)
```
